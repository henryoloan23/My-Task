```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Membaca data dari file Excel
data = pd.read_excel('/content/data_quiz.xlsx', sheet_name='Gabungan')
print(data)

```

```python
# 1. Boxplot dari nilai kuis grup by kelas
plt.figure(figsize=(10, 6))
sns.boxplot(x='Kelas', y='Quiz', data=data)
plt.title('Boxplot Nilai Kuis by Kelas')
plt.show()
```

```python
# 2. Boxplot dari nilai kuis grup by gender
plt.figure(figsize=(10, 6))
sns.boxplot(x='Gender', y='Quiz', data=data)
plt.title('Boxplot Nilai Kuis by Gender')
plt.show()
```

```python
# 3. Boxplot dari nilai IPS grup by kelas
plt.figure(figsize=(10, 6))
sns.boxplot(x='Kelas', y='IPS2', data=data)
plt.title('Boxplot Nilai IPS by Kelas')
plt.show()
```

```python
# 4. Boxplot dari nilai IPS grup by gender
plt.figure(figsize=(10, 6))
sns.boxplot(x='Gender', y='IPS2', data=data)
plt.title('Boxplot Nilai IPS by Gender')
plt.show()
```

```python
# 5. Scatter plot antara nilai kuis dengan nilai IPS
plt.figure(figsize=(10, 6))
sns.scatterplot(x='IPS2', y='Quiz', data=data)
plt.title('Scatter Plot Nilai Kuis vs Nilai IPS')
plt.show()
```

```python
# 6. Cari nilai korelasi antara nilai kuis dengan nilai IPS
correlation = data['IPS2'].corr(data['Quiz'])
print(f'Nilai Korelasi antara Nilai Kuis dan Nilai IPS: {correlation:.2f}')
```
